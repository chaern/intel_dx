```python
# import pymodbus
from pymodbus.client import ModbusTcpClient
import time as tt
import threading
ip = '192.168.0.16'
client = ModbusTcpClient(ip, port=502)
client.connect()
```

```python
Fvalve= 10 * 100
client.write_register(0, Fvalve )
```

```python
Dvalve = 10 * 100
client.write_register(1, Dvalve )
```

```python
client.write_registers(0, [1000,0] ) # 여러개의 값을 쓰러면 뒤에 배열로 넣기
```

```python
scan_run = True
target = 1000
def meter_scan():
    while scan_run:
        water_level = client.read_input_registers(0, count = 4).registers
        print(f'\r스캔시작 물의 높이: {water_level[0]/10:.1f}%  ',end="")

        if (target - water_level[0])<10:
            client.write_register(0,0)
        elif (target - water_level[0])<100:
            client.write_register(0,100)
        elif (target - water_level[0])<300:
            client.write_register(0,200)
        elif (target - water_level[0])<500:
            client.write_register(0,500)
        elif (target - water_level[0])<700:
            client.write_register(0,800)
        elif (target - water_level[0])<=1000:
            client.write_register(0,1000)

        tt.sleep(0.1)

scan = threading.Thread(target=meter_scan, daemon=True)
print('모니터링 시작')
scan.start()
```

```python
print("모니터링 정지")
scan_run = False
```
